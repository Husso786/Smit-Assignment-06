## A traffic system calculates fines:

Speed limit = 60 km/h
* If speed ≤ 60 → No fine
* If 61–80 → Fine = 500
* If 81–100 → Fine = 2000
* If >100 → License suspended
#### BUT:
*If emergency vehicle → no fine
*If rainy weather AND speed > 60 → fine increases by 50%

```python
speed = int(input('Enter speed (km/h)'))
emergency = input('Is it an emergency vehicle? (yes/no)')
rainy = input('Is the weather rainy? (yes/no)')

if emergency == 'yes':
    print('No fine')
else:
    if speed <= 60:
        print('No fine')
    elif speed <= 80:
        fine = 500
        if rainy == 'yes':
            fine = fine * 1.5
        print('Fine =', fine)
    elif speed <= 100:
        fine = 2000
        if rainy == 'yes':
            fine = fine * 1.5
        print('Fine =', fine)
    else:
        print('License suspended')
```

## A company filters candidates:
* Experience ≥ 5 years AND skill score ≥ 80 → shortlisted
* Experience 2–4 years AND skill score ≥ 70 → interview
* Freshers → only if test score ≥ 85
* If criminal record = True → reject immediately
* If multiple job switches (>3 in 2 years) → flag as unstable

```python
experience = int(input('Experience in years'))
skill_score = int(input('Skill score'))
test_score = int(input('Test score'))
criminal_record = input('Criminal record (True/False)')
job_switches = int(input('Job switches in 2 years'))
if experience > 5 and skill_score > 80:
    print('Shortlisted')
elif experience >= 2 and experience <= 4 and skill_score >= 70:
    print('Interview')
elif test_score > 85:
    print('Fresher')
elif criminal_record == 'True':
    print('Reject Immediately')
elif job_switches > 3:
    print("Flag as unstable")
else:
    print('Not Selected')
```

## Write a program to classify a person’s risk level in a police database.
Take input:
* criminal_record (Yes/No)
* number_of_cases
* travel_history_flagged (Yes/No)
* financial_activity_score
* last_activity_days
#### Rules:
* criminal_record = Yes AND cases > 3 → High Risk
* flagged travel + suspicious financial score → High Risk
* no activity for long time (>365 days) → Monitor
* clean record + normal activity → Low Risk
* mixed signals → Medium Risk

```python
criminal_record = input('Criminal record? (yes/no)')
number_of_cases = int(input('cases'))
travel_history_flagged = input('Travel flagged? ')
financial_activity_score = int(input('Financial score'))
last_activity_days = int(input('Last activity days'))

if criminal_record == 'yes' and number_of_cases > 3:
    print('High Risk')
elif travel_history_flagged == 'yes' and financial_activity_score  > 50:
    print('High Risk')
elif last_activity_days > 365:
    print('Monitor')
elif criminal_record == 'no' and travel_history_flagged == 'no':
    print('Low Risk')
else:
    print('Medium Risk')
```
